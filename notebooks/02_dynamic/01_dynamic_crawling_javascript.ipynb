## 원하는 요소가 생성될 때까지 대기 후 검색

동적 페이지에서는 페이지 접속 직후 데이터를 찾지 않고,  
`WebDriverWait`을 사용하여 원하는 요소가 생성될 때까지 기다린다.

```python
from selenium.webdriver.support import expected_conditions as EC
```

```python
## 동적 요소 최대 대기 시간(초)
WAIT_TIMEOUT = 10

driver = webdriver.Chrome()
driver.get(TARGET_URL)

wait = WebDriverWait(driver, WAIT_TIMEOUT)

## 첫 번째 인용문 요소가 DOM에 생성될 때까지 대기
wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, '.quote')))

print(f'페이지 제목 : {driver.title}')
print(f'현재 URL : {driver.current_url}')

driver.quit()
print('웹브라우저 종료')
```

## [함수 정의] 인용문 요소 한 건을 딕셔너리로 변환

```python
def parse_quote_element(
    quote_element: WebElement,
    page: int, 
    source_url: str,    
) -> dict[str, str | int]:
    """
    인용문 HTML 요소 한 건에서
    인용문, 저자, 태그 정보를 추출한다.    
    """
    quote_text = quote_element.find_element(By.CSS_SELECTOR, '.text').text
    author = quote_element.find_element(By.CSS_SELECTOR, '.author').text
    tag_elements = quote_element.find_elements(By.CSS_SELECTOR, '.tags .tag')
    tags = [tag.text for tag in tag_elements]

    return {
        'quote': quote_text,
        'author': author,
        'tags': '|'.join(tags),
        'page': page,
        'source_url': source_url,
    }   
```

```python
quote_elements[0]
```

```python
driver = webdriver.Chrome()
driver.get(TARGET_URL)

wait = WebDriverWait(driver, WAIT_TIMEOUT)
wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, '.quote')))
quote_elements = driver.find_elements(By.CSS_SELECTOR, '.quote')

parse_quote_element(quote_elements[0], 1, driver.current_url)
```

```python
driver.quit()
print('웹브라우저 종료')
```

## 현재 페이지 전체 데이터 추출 함수

```python
len(quote_elements)
```

```python
def collect_current_page(
    driver: webdriver.Chrome,
    wait: WebDriverWait,
    page: int,
) -> list[dict[str, str | int]]:
    """
    현재 브라우저 페이지의 모든 인용문을 수집한다.
    """
    quote_elements = wait.until(
        EC.presence_of_all_elements_located([By.CSS_SELECTOR, '.quote'])
    )

    source_url = driver.current_url

    return [
        parse_quote_element(quote_element, page, source_url)
        for quote_element 
        in quote_elements
    ]
```

```python
## 동적 요소 최대 대기 시간(초)
WAIT_ITMEOUT = 10

driver = webdriver.Chrome()
driver.get(TARGET_URL)

wait = WebDriverWait(driver, WAIT_ITMEOUT)
# wait.until(EC.presence_of_element_located([By.CSS_SELECTOR, '.quote']))
# quote_elements = driver.find_elements(By.CSS_SELECTOR, '.quote')

page_1_quotes = collect_current_page(driver, wait, 1)
print(f'1페이지 수집 데이터 수 : {len(page_1_quotes)}')

```
